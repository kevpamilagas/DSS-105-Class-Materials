# Examples of editing/creating columns

```python
store_cleaned['Date'] =  store_cleaned['Trans_id'].str[5:10].astype(int) #you'll have to tell them that the first date is in fact January 4th
store_cleaned['Time'] = store_cleaned['Trans_id'].str[10:].astype(float)
store_cleaned['Store'] = store_cleaned['Trans_id'].str[:4]
store_cleaned.head(2)
```

```python
store_cleaned['Stores'] = store_cleaned['Trans_id'].str[0:4:]
store_cleaned = store_cleaned.drop('Stores', axis=1)

store_cleaned.head(2)
```

```python
from datetime import datetime, timedelta

store_cleaned['DateTime'] = datetime(1900, 1, 1) + pd.to_timedelta(store_cleaned['Date'] + store_cleaned['Time'] - 2, unit='D').dt.round('s')
store_cleaned['Month'] = store_cleaned['DateTime'].dt.month_name()
store_cleaned.head(2)
```

```python
time = 1100

if time >= 1000 and time < 1100:
    print('Morning')
elif time >= 1100 and time < 1400:
    print('Lunch')
elif time >= 1400 and time < 1600:
    print('Mid-Day')
elif time >= 1600 and time < 2100:
    print('Dinner')
elif time >= 2100 and time <= 2200:
    print('Evening')
else:
    print('Error')
```

```python
def shift_name(time):
    if time < 1000/2400 or time >= 2200/2400:
        return 'Error'
    elif time < 1100/2400:
        return 'Morning'
    elif time < 1400/2400:
        return 'Lunch'
    elif time < 1600/2400:
        return 'Mid-Day'
    elif time < 2100/2400:
        return 'Dinner'
    else:
        return 'Evening'
```

```python
store_cleaned['Shift'] = store_cleaned['Time'].apply(shift_name)
store_cleaned.head(2)
```

# Pandas Mapping

```python
sizes = {
    1: '6-inch',
    2: '12-inch'
}

sandwiches = {
    1: 'Cold Cut', 
    2: 'BLT',
    3: 'Italian',
    4: 'Meatball',
    5: 'Chicken',
    6: 'Steak',
    7: 'Pastrami'
}
```

```python
store_cleaned['Size'] = store_cleaned['Sand_id'].map(sizes)
store_cleaned['Sand'] = store_cleaned['Menu_id'].map(sandwiches)
store_cleaned.head(2)
```
