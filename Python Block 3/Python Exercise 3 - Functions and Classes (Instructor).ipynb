## 2. Create a `Car` class with `make`, `model`, and `year` attributes. Create an instance of your car and print its `year`.

```python
class Car:
    def __init__(self, car_make, car_model, car_year):
        self.make = car_make
        self.model = car_model
        self.year = car_year
```

```python
my_car = Car('Toyota', 'Camry', 2024)
print(my_car.year)
```

```python
#show type and other attributes
print(type(my_car))
print(my_car.make)
print(my_car.model)
```

## 3. (Bonus) Add a `describe()` method to your `Car` class that prints a description using its attributes. (e.g., `"2024 Toyota Camry"`)

```python
class Car:
    def __init__(self, car_make, car_model, car_year):
        self.make = car_make
        self.model = car_model
        self.year = car_year

    def describe(self):
        print(str(self.year) + ' ' + self.make + ' ' + self.model)
```

```python
#remember to re-run the class AND create a new instance
my_car = Car('Toyota', 'Camry', 2024)
my_car.describe()
```

```python
#alternative using f-string
class Car:
    def __init__(self, car_make, car_model, car_year):
        self.make = car_make
        self.model = car_model
        self.year = car_year

    def describe(self):
        print(f'{self.year} {self.make} {self.model}')

my_car = Car('Toyota', 'Camry', 2024)
my_car.describe()
```
